# Tokenization and Normalization

**Workshop stage:** 2 / preprocessing  
**Main skill:** compare raw and processed Arabic text  
**Tool:** CAMeL Tools

This notebook introduces two common preprocessing steps in Arabic NLP: tokenization and normalization. These steps make text easier to process, but they also involve interpretive choices.

## Tokenization Limits

Tokenization means dividing text into units, often words or punctuation marks. Splitting on spaces is a useful beginning, but Arabic morphology can attach particles, prefixes, suffixes, and clitics to written forms.

```python
sentence = 'والولدُ قرأ الكِتابَ في المدرسة.'
print('Original sentence:', sentence)
print('Whitespace tokens:', sentence.split())
```

## Arabic-Aware Tokenization

CAMeL Tools provides functions designed for Arabic NLP. The next cell uses a simple tokenizer that separates punctuation more carefully than plain `split()`.

| Method | Strength | Limit |
| --- | --- | --- |
| `split()` | Easy to understand | Only sees spaces |
| `simple_word_tokenize()` | Handles punctuation better | Still not a full morphological analysis |

```python
from camel_tools.tokenizers.word import simple_word_tokenize

text = 'والولدُ قرأ الكِتابَ في المدرسة.'
tokens = simple_word_tokenize(text)
print('CAMeL Tools tokens:', tokens)
```

## Normalization Example

Normalization converts text into a more standard form. This can help with searching and counting, but it can also remove information such as diacritics.

```python
from camel_tools.utils.dediac import dediac_arabic
from camel_tools.utils.normalize import normalize_alef_maksura

text = 'هٰذَا نَصٌّ بِالْعَرَبِيَّةِ'
without_diacritics = dediac_arabic(text)
normalized = normalize_alef_maksura(without_diacritics)

print('Original:', text)
print('Without diacritics:', without_diacritics)
print('Normalized:', normalized)
```

## Why This Matters

| Benefit | Risk |
| --- | --- |
| Searching becomes more consistent | Meaningful variation may be flattened |
| Counts become easier to compare | Editorial or linguistic detail may disappear |
| Models receive cleaner input | The cleaning choices may encode assumptions |
