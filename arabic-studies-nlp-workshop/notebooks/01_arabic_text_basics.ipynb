## Python Strings and Unicode

A **string** is a piece of text inside quotation marks. Python stores text using Unicode, a standard that gives characters from many writing systems their own codes.

| Function | Question it answers |
| --- | --- |
| `print(text)` | What does the text look like? |
| `len(text)` | How many characters does Python count? |
| `text.split()` | What happens if we split at spaces? |
| `list(text)` | Which individual characters are present? |

```python
text = 'مرحبا بالعالم'
print(text)
print('Characters counted by len():', len(text))
print('Words from split():', text.split())
print('Characters as a list:', list(text))
```

## What to Notice

`len()` counts characters, including spaces. `split()` is simpler: it separates text wherever it sees whitespace. This is useful, but it is not the same as a full linguistic analysis of Arabic words.

> **Academic caution:** a computer count is exact within its rules, but the rules may not match the scholarly category you care about.

## Working With Diacritics

Some visible marks are not independent letters. Arabic diacritics can affect meaning and pronunciation, and they also affect what a computer counts.

```python
diacritics_text = 'هٰذَا نَصٌّ بِلُغَةٍ عَرَبِيَّةٍ'
print(diacritics_text)
print('Words from split():', diacritics_text.split())
print('Characters and diacritics:', list(diacritics_text))
```
