## Before Running a Model

| Question | Why it matters |
| --- | --- |
| What data was used to train the model? | Training data shapes predictions |
| Which dialects, genres, periods, and registers are missing? | Missing data creates blind spots |
| Is the model predicting or verifying? | Prediction is not proof |
| What would count as responsible evidence in your field? | Scholarly claims need context |

## Masked Language Modeling

A masked language model receives a sentence with one missing piece and predicts likely replacements. This can be useful for demonstration, but it is not the same as understanding historical context, authorial intention, or truth.

```python
model_name = 'aubmindlab/bert-base-arabertv02'
sample = 'أريد أن أزور [MASK] اليوم'
print('Model selected:', model_name)
print('Input sentence:', sample)
```

## Optional Live Demo

> **If the internet is slow:** skip this cell and discuss the concept from the markdown. The workshop does not depend on every participant downloading a model.

The next cell may download model files the first time it runs.

```python
try:
    from transformers import AutoTokenizer, AutoModelForMaskedLM, pipeline

    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForMaskedLM.from_pretrained(model_name)
    masker = pipeline('fill-mask', model=model, tokenizer=tokenizer)

    predictions = masker(sample)
    print('Input:', sample)
    for result in predictions[:5]:
        print(result['token_str'], round(result['score'], 4))
except Exception as error:
    print('The live model demo could not run in this environment.')
    print('This is acceptable for the workshop.')
    print('Reason:', error)
```

## Interpreting the Output

| Output feature | Responsible interpretation |
| --- | --- |
| Ranked words | Likely completions according to the model |
| Probability score | Confidence within the model, not truth |
| Surprising prediction | A prompt for investigation, not an error by default |
| Missing dialect or genre | Possible training-data limitation |

> **Academic caution:** model output is computational evidence requiring context, not an independent scholarly conclusion.
