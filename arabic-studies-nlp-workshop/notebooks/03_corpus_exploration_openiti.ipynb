## Corpus Workflow

| Step | Question |
| --- | --- |
| Load text | What is in the file? |
| Inspect metadata | Where does the text come from? |
| Count words | Which forms occur often? |
| Search context | How does a word appear in use? |
| Reflect | What can and cannot be claimed? |

## Load a Small Corpus Sample

`Path` helps Python find files. The sample text lives in `data/raw/`.

```python
from pathlib import Path

sample_path = Path('../data/raw/openiti_sample.txt')
if not sample_path.exists():
    sample_path = Path('data/raw/openiti_sample.txt')

text = sample_path.read_text(encoding='utf-8')
print(text)
```

## Inspect the Metadata

Metadata describes the text. It may include author, title, date, genre, language, source, or editorial information. Corpus analysis is weaker when metadata is missing or misunderstood.

```python
metadata_path = Path('../data/metadata/sample_metadata.csv')
if not metadata_path.exists():
    metadata_path = Path('data/metadata/sample_metadata.csv')

metadata = metadata_path.read_text(encoding='utf-8')
print(metadata)
```

## Frequency Counts

A frequency count asks: which written forms occur most often in this sample? This is a descriptive result, not an interpretation by itself.

```python
words = [word.strip(' .،؛:؟!') for word in text.split()]
words = [word for word in words if word]

freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1

for word, count in sorted(freq.items(), key=lambda item: item[1], reverse=True):
    print(word, count)
```

## Concordance Idea

A concordance shows a word near its surrounding context. This helps us move from counting back toward reading.

```python
query = 'النصوص'
position = text.find(query)

if position == -1:
    print('The query was not found:', query)
else:
    start = max(position - 40, 0)
    end = min(position + len(query) + 40, len(text))
    print(text[start:end])
```

## Text Quality and Corpus Limits

| Issue | Why it matters |
| --- | --- |
| OCR errors | A count may reflect recognition mistakes |
| Transcription differences | The same word may appear in multiple forms |
| Metadata uncertainty | Interpretation depends on provenance |
| Genre imbalance | Visible language is shaped by corpus selection |
| Dialect absence | Models and counts may privilege certain registers |

> **Academic caution:** frequency is a starting point for interpretation, not the interpretation itself.
