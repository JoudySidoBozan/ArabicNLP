## Variables

A variable is a name that stores a value.

| Code | Meaning |
| --- | --- |
| `message = ...` | Store text under the name `message` |
| `print(message)` | Show the stored text |

```python
message = 'Notebook navigation example: top to bottom, cell by cell.'
print(message)
```
